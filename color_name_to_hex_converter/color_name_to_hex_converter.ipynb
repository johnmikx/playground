```python
from webcolors import name_to_hex

def color_name_to_hex(color_name):
  try:
    color_hex = name_to_hex(color_name)
    return color_hex
  except ValueError:
    return None

colorname = input("Enter color name: ")
result_code = color_name_to_hex(colorname)
print(result_code)
```
